#Draw elbow plot and from that figure out optimal value of k

```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import pandas as pd
from sklearn import datasets
```

# Load iris Data set

```python
iris = datasets.load_iris()
iris_df = pd.DataFrame(iris.data, columns=iris.feature_names)
```

```python
# Keep only the petal length and petal width features
x = iris_df[['petal length (cm)', 'petal width (cm)']]

# Apply MinMaxScaler to scale the features between 0 and 1
scaler=MinMaxScaler()
x_scaled = scaler.fit_transform(x)

# Create a list to store the WCSS (Within-Cluster Sum of Squares)
wcss = []

# Loop over different values of k to calculate WCSS
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k)
    kmeans.fit(x_scaled)
    wcss.append(kmeans.inertia_)

# Plot the WCSS values to visualize the elbow
plt.plot(range(1, 11), wcss, marker='o')
plt.title('Elbow Plot for Optimal k')
plt.xlabel('Number of clusters (k)')
plt.ylabel('WCSS (Inertia)')
plt.grid(True)
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import pandas as pd
from sklearn import datasets

# Load the iris dataset
iris = datasets.load_iris()
iris_df = pd.DataFrame(iris.data, columns=iris.feature_names)

# Keep only the petal length and petal width features
X = iris_df[['petal length (cm)', 'petal width (cm)']]

# Apply StandardScaler to scale the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Create and fit the KMeans model with 3 clusters
kmeans = KMeans(n_clusters=3)
kmeans.fit(X_scaled)

# Get the predicted cluster labels and centroids
labels = kmeans.labels_
centroids = kmeans.cluster_centers_

# Plot the clustered data
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=labels, cmap='rainbow')
plt.scatter(centroids[:, 0], centroids[:, 1], color='black', marker='*', s=200)
plt.title('Clusters of Iris Flowers (After Scaling)')
plt.xlabel('Standardized Petal Length')
plt.ylabel('Standardized Petal Width')
plt.show()
```
